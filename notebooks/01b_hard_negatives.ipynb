```python
%pip install faker pandas
import random
import pandas as pd
from faker import Faker

fake = Faker(); random.seed(7); Faker.seed(7)
R = random.randint
```

```python
def date():
    y, m, d = R(2019, 2026), R(1, 12), R(1, 28)
    return random.choice([f"{y}-{m:02d}-{d:02d}", f"{m:02d}/{d:02d}/{y}", f"{d:02d}.{m:02d}.{y}",
                          f"{d}/{m}/{y}", f"{y}/{m:02d}/{d:02d}", fake.date(pattern="%d %b %Y"),
                          fake.date(pattern="%B %d, %Y")])

def time_():
    return random.choice([f"{R(1,12)}:{R(0,59):02d} {random.choice(['AM','PM'])}",
                          f"{R(0,23):02d}:{R(0,59):02d}", f"{R(0,23):02d}:{R(0,59):02d}:{R(0,59):02d}"])

def money():
    amt = f"{R(1,9999):,}.{R(0,99):02d}"
    return random.choice([f"${amt}", f"${R(1,999)}.{R(0,99):02d}", f"Rs. {amt}", f"USD {amt}",
                          f"{amt} USD", f"${R(1,99999):,}"])

def ident():
    pre = random.choice(["INV", "CLM", "ACCT", "ORD", "REF", "TKT", "PO", "SKU", "ID", "CASE", "REQ", "TXN"])
    return random.choice([f"{pre}-{R(1000,999999)}", f"{pre}-{R(10,99)}-{R(100000,999999)}",
                          f"{pre}{R(1000,99999)}", f"#{R(1000,99999)}", f"{pre}-{R(2019,2026)}-{R(1,9999):04d}"])

def code():
    return random.choice([str(R(10000, 99999)), str(R(1000, 99999)),
                          f"{random.choice('ABCDEFGHJKLMNPQRSTUVWXYZ')}{R(10,99)}.{R(0,9)}{random.choice(['XXA','XXD','A','B',''])}",
                          f"M{R(10,99)}.{R(0,9)}", f"CPT {R(10000,99999)}",
                          f"ICD-10 {random.choice('ABCDEFGHJKLMNPQRS')}{R(10,99)}.{R(0,99)}"])

def label_only():
    return random.choice(["Name:", "Phone:", "Email:", "Date:", "With:", "When:", "Patient:", "Status:", "Type:",
        "Account #:", "Invoice #:", "Regarding:", "Bill To:", "Adjuster:", "Team:", "Assigned:", "Due:", "DOB:",
        "Claimant", "Callback", "Summary", "Date of Service", "Billed Amount", "Qty", "Description / Diagnosis",
        "Place of Service", "Total Billed:", "Amount Due:", "Password", "Username", "Login", "Contact", "Address:",
        "Reference No.", "Notes", "Settings", "Reports", "Payments", "History", "Documents", "Coverage"])

def sentence_ui():
    return random.choice(["Thank you for your prompt payment", "Please remit payment within 30 days.",
        "Reviewed provider invoice and EOB.", "Waiting on patient statement.", "Treatment plan follow up",
        "Page 1 of 2", f"Page {R(1,9)} of {R(2,12)}", "Follow-up visit", "Office visit, established patient",
        "Physical therapy evaluation", "Emergency room visit, level 3", "Type here to search",
        "Approve & Issue Payment", "Close Claim", "Add Note", "Total Reserve: $0.00", "Insurance Adjustment:",
        "Coverage Verified", "Documents Received", "Decision Pending", "Under Review", "Pending Review",
        "Your password must be at least 12 characters", "Forgot your password?", "Enter your phone number",
        "Please confirm your email address", "We will call you back soon", "Reply to this email for help"])

def make_hard():
    k = random.choice(["date", "date", "date_time", "datetime_text", "money", "ident", "ident", "code",
                       "label", "label", "ui", "ui", "label_date", "label_money", "label_id",
                       "range", "count", "percent"])
    if k == "date": return date()
    if k == "date_time": return f"{date()} {time_()}"
    if k == "datetime_text": return f"{date()} {time_()} - {fake.name()}"
    if k == "money": return money()
    if k == "ident": return ident()
    if k == "code": return code()
    if k == "label": return label_only()
    if k == "ui": return sentence_ui()
    if k == "label_date":
        return f"{random.choice(['Date:', 'Due:', 'Assigned:', 'DOB', 'Date of Loss:', 'Invoice Date:'])} {date()}"
    if k == "label_money":
        return f"{random.choice(['Total:', 'Amount Due:', 'Balance:', 'Paid:'])} {money()}"
    if k == "label_id":
        return f"{random.choice(['Claim:', 'Invoice #:', 'Account #:', 'Order', 'Ref:'])} {ident()}"
    if k == "range":
        return f"Date(s) of Service: {date()} to {date()}"
    if k == "count":
        return f"{R(1,99)} items, {R(1,50)} of {R(50,200)}"
    return f"{R(1,100)}%  {R(1,9999):,} units"
```

```python
rows = [{"text": make_hard(), "label": "safe", "source": "hard_negatives"} for _ in range(4000)]
df = pd.DataFrame(rows).drop_duplicates(subset="text").reset_index(drop=True)
df.to_csv("../data/hard_negatives.csv", index=False)
print(df.shape)
df.sample(20, random_state=1)
```
